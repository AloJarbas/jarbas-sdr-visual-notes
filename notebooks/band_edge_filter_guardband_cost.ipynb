# Band-edge filter guardband cost

This notebook is the slower companion to `assets/2026-05-22-band-edge-guardband-cost-comparison.*`.

The narrow question is simple:

**once the half-sine band-edge design fixes the moderate-roll-off slope, how much adjacent-channel footprint did that fix buy?**


![Band-edge guardband cost comparison](../assets/2026-05-22-band-edge-guardband-cost-comparison.png)

## Scope

This notebook stays inside one bounded detector-path study.
It does **not** claim to be a full adjacent-channel rejection benchmark or a closed-loop receiver result.

It only measures what the lower + upper band-edge filters themselves keep capturing.


## The two metrics

For each filter family, the notebook tracks:

### 1. Near-lock slope

Using the same central finite difference from the previous note,

\[
S \approx \frac{D(+\Delta f)-D(-\Delta f)}{2\Delta f},
\]

with \(\Delta f = 0.01 R_s\).

### 2. Adjacent-only capture ratio

For a neighboring QPSK waveform centered at spacing \(\Delta f_{adj}\),

\[
C_{adj} = \frac{E_{lower}+E_{upper}}{E_{input}}.
\]

This is the study's selectivity proxy.
It asks how much adjacent energy the band-edge detector path keeps seeing.


## The key 63-tap table

| roll-off `α` | design | slope | pickup at `1.0 R_s` | spacing for `≤ 5%` pickup |
|---|---|---:|---:|---:|
| `0.20` | proxy | `0.480` | `4.2%` | `0.75 R_s` |
| `0.20` | half-sine | `0.943` | `18.3%` | `1.20 R_s` |
| `0.35` | proxy | `0.670` | `10.7%` | `1.15 R_s` |
| `0.35` | half-sine | `0.986` | `31.5%` | `1.35 R_s` |
| `0.50` | proxy | `0.764` | `17.3%` | `1.25 R_s` |
| `0.50` | half-sine | `1.003` | `45.0%` | `1.55 R_s` |

That table is the whole reason this follow-up exists.


## Main read

### The half-sine correction is real

The old proxy really was understating the moderate-roll-off near-lock slope.
That part of the previous note survives unchanged.

### The price is also real

At `α = 0.35` and `63` taps, the slope rises from `0.670` to `0.986`.
But the adjacent pickup at `1.0 R_s` jumps from `10.7%` to `31.5%`.
The spacing needed to get below `5%` pickup widens from `1.15 R_s` to `1.35 R_s`.

So the repo can stop calling the half-sine design a free fix.
It is a tradeoff.


## One useful surprise: tap count mostly changes slope, not spacing threshold

At `α = 0.35`:

- proxy slope goes from `0.670` (`63` taps) to `0.983` (`255` taps),
- but the `≤ 5%` spacing stays at `1.15 R_s`.

For the half-sine family:

- slope goes from `0.986` to `1.042`,
- but the spacing stays at `1.35 R_s`.

That is a sharper result than “wider filters hear more.”
In this bounded setup, the footprint is being set mostly by the filter family itself.


## Adversarial check

A weak version of this follow-up would have stopped at another slope plot and called the problem solved.

That would have been incomplete.
The new design really is better near lock, but it is also materially less selective on this adjacent-only metric.

Without that second measurement, the public story would still be too flattering.


## What this notebook does **not** prove

- It does not prove overall receiver performance under a real adjacent interferer power ratio.
- It does not prove loop stability or BER.
- It does not prove a universal guardband rule for every band-edge implementation.

It only closes the specific loophole left by the previous repo state.


## Best next move

If this branch gets one more pass, the strongest next step is one of these:

- put the same slope-versus-selectivity tradeoff into a closed-loop simulation,
- or sweep adjacent interferer level explicitly instead of assuming a unit-power neighbor.

Anything smaller would probably just redraw the same point.
